```python
import shutil
from pathlib import Path
import pandas as pd

# ==============================================================================
# CONFIGURATION: Point this to the folder containing your extracted year folders
# ==============================================================================
EXTRACTED_ROOT_DIR = "."                # Folder containing 2010/, 2015/, 2017/, 2018/, 2021/
OUTPUT_DATASET_DIR = "./milco_dataset"  # Where the clean YOLO dataset will be built
COPY_CLEAN_DATASET = True               # Set True to copy & organize files for YOLO

YEARS = ["2010", "2015", "2017", "2018", "2021"]
IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}
IGNORE_TXTS = {"classes.txt", "obj.names", "obj.data", "train.txt", "test.txt"}


def is_valid_path(p: Path) -> bool:
    """Ignore macOS hidden files (__MACOSX, .DS_Store, ._*) and hidden folders."""
    return not any(
        part.startswith("__MACOSX") or part.startswith("._") or part.startswith(".")
        for part in p.parts
    )


def find_year_folder(root: Path, year: str, exclude_dir: Path) -> Path:
    """Locate the extracted folder for a given survey year."""
    direct = root / year
    if direct.is_dir():
        return direct
    # Search one or two levels deep if folders were extracted inside a parent directory
    for candidate in root.rglob(year):
        if candidate.is_dir() and is_valid_path(candidate):
            if exclude_dir.resolve() not in candidate.resolve().parents and candidate.resolve() != exclude_dir.resolve():
                return candidate
    raise FileNotFoundError(f"Could not find extracted folder for year '{year}' inside {root.resolve()}")


def parse_yolo_txt(txt_path: Path):
    """Read a YOLO .txt annotation file and return (milco_count, nombo_count, valid_lines)."""
    milco, nombo = 0, 0
    valid_lines = []
    if txt_path is not None and txt_path.is_file():
        text = txt_path.read_text(encoding="utf-8", errors="ignore")
        for line in text.splitlines():
            line = line.strip()
            if not line:
                continue
            parts = line.split()
            if len(parts) >= 5 and parts[0].isdigit():
                cls_id = int(parts[0])
                if cls_id == 0:
                    milco += 1
                elif cls_id == 1:
                    nombo += 1
                valid_lines.append(line)
    return milco, nombo, valid_lines


def build_dataset_from_folders(root_dir: str, output_dir: str, copy_files: bool = True):
    root = Path(root_dir)
    out_dir = Path(output_dir)
    all_records = []

    for year in YEARS:
        year_dir = find_year_folder(root, year, out_dir)

        img_map = {}
        txt_map = {}

        # Scan all files inside the year folder
        for f in year_dir.rglob("*"):
            if not f.is_file() or not is_valid_path(f):
                continue
            ext = f.suffix.lower()
            stem = f.stem
            if ext in IMG_EXTS:
                img_map[stem] = f
            elif ext == ".txt" and f.name.lower() not in IGNORE_TXTS:
                txt_map[stem] = f

        if copy_files:
            img_out_dir = out_dir / "images" / year
            lbl_out_dir = out_dir / "labels" / year
            img_out_dir.mkdir(parents=True, exist_ok=True)
            lbl_out_dir.mkdir(parents=True, exist_ok=True)

        for stem, img_path in sorted(img_map.items()):
            txt_path = txt_map.get(stem, None)
            milco, nombo, valid_lines = parse_yolo_txt(txt_path)

            total_objs = milco + nombo
            is_empty = (total_objs == 0)

            # Prefix with year so filenames are globally unique across the dataset
            clean_stem = f"{year}_{stem}" if not stem.startswith(year) else stem
            img_ext = img_path.suffix.lower()

            dest_img = out_dir / "images" / year / f"{clean_stem}{img_ext}"
            dest_lbl = out_dir / "labels" / year / f"{clean_stem}.txt"

            if copy_files:
                shutil.copy2(img_path, dest_img)
                dest_lbl.write_text(
                    "\n".join(valid_lines) + ("\n" if valid_lines else ""),
                    encoding="utf-8"
                )

            all_records.append({
                "YEAR": year,
                "IMAGE_NAME": f"{clean_stem}{img_ext}",
                "SOURCE_IMAGE": str(img_path.resolve()),
                "IMAGE_PATH": str(dest_img.resolve()) if copy_files else str(img_path.resolve()),
                "LABEL_PATH": str(dest_lbl.resolve()) if copy_files else (str(txt_path.resolve()) if txt_path else ""),
                "EMPTY": int(is_empty),
                "MILCO": milco,
                "NOMBO": nombo,
                "TOTAL_OBJECTS": total_objs
            })

    df_images = pd.DataFrame(all_records)

    # Build the per-year summary rows
    summary_rows = []
    for year in YEARS:
        sub = df_images[df_images["YEAR"] == year]
        n_imgs = len(sub)
        n_empty = int(sub["EMPTY"].sum()) if n_imgs > 0 else 0
        n_milco = int(sub["MILCO"].sum()) if n_imgs > 0 else 0
        n_nombo = int(sub["NOMBO"].sum()) if n_imgs > 0 else 0
        avg_obj = (n_milco + n_nombo) / n_imgs if n_imgs > 0 else 0.0

        summary_rows.append({
            "YEAR": year,
            "IMAGES": n_imgs,
            "EMPTY": n_empty,
            "MILCO": n_milco,
            "NOMBO": n_nombo,
            "AVG_OBJECTS": round(avg_obj, 2)
        })

    df_summary = pd.DataFrame(summary_rows)

    # Calculate TOTAL row
    tot_imgs = int(df_summary["IMAGES"].sum())
    tot_empty = int(df_summary["EMPTY"].sum())
    tot_milco = int(df_summary["MILCO"].sum())
    tot_nombo = int(df_summary["NOMBO"].sum())
    tot_avg = round((tot_milco + tot_nombo) / tot_imgs, 2) if tot_imgs > 0 else 0.0

    # Print plain-text table matching the exact screenshot layout
    header = f"{'YEAR':<8}{'IMAGES':<9}{'EMPTY':<8}{'MILCO':<8}{'NOMBO':<8}{'AVG_OBJECTS':<12}"
    print(header)
    for r in summary_rows:
        print(f"{r['YEAR']:<8}{r['IMAGES']:<9}{r['EMPTY']:<8}{r['MILCO']:<8}{r['NOMBO']:<8}{r['AVG_OBJECTS']:<12.2f}")
    print("-" * 53)
    print(f"{'TOTAL':<8}{tot_imgs:<9}{tot_empty:<8}{tot_milco:<8}{tot_nombo:<8}{tot_avg:<12.2f}")

    # Append TOTAL row to the summary DataFrame
    df_total = pd.DataFrame([{
        "YEAR": "TOTAL",
        "IMAGES": tot_imgs,
        "EMPTY": tot_empty,
        "MILCO": tot_milco,
        "NOMBO": tot_nombo,
        "AVG_OBJECTS": tot_avg
    }])
    df_summary_full = pd.concat([df_summary, df_total], ignore_index=True)

    # Save CSVs and Ultralytics YOLO configuration file
    if copy_files and len(df_images) > 0:
        out_dir.mkdir(parents=True, exist_ok=True)
        df_summary_full.to_csv(out_dir / "dataset_year_summary.csv", index=False)
        df_images.to_csv(out_dir / "dataset_image_index.csv", index=False)

        yaml_content = f"""# MILCO/NOMBO Year-Disjoint Dataset Configuration
path: {out_dir.resolve()}
train:
  - images/2015
  - images/2017
  - images/2021
val:
  - images/2010
  - images/2018
test:
  - images/2010
  - images/2018

nc: 2
names:
  0: MILCO
  1: NOMBO
"""
        (out_dir / "milco_year_disjoint.yaml").write_text(yaml_content, encoding="utf-8")
        print(f"\n[✓] Clean dataset, CSVs, and YOLO config saved to: {out_dir.resolve()}")

    return df_summary_full, df_images


# Run the script:
summary_df, images_df = build_dataset_from_folders(
    EXTRACTED_ROOT_DIR,
    OUTPUT_DATASET_DIR,
    COPY_CLEAN_DATASET
)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd

# ==============================================================================
# CONFIGURATION
# ==============================================================================
DATASET_DIR = Path("./milco_dataset")
VAL_RATIO = 0.15   # 15% of TRAIN years held out for Validation (85% Train / 15% Val)
RANDOM_SEED = 42   # Fixed seed for 100% reproducible splits

df_images = pd.read_csv(DATASET_DIR / "dataset_image_index.csv")
df_images["YEAR"] = df_images["YEAR"].astype(str)


def stratified_year_split(df_pool: pd.DataFrame, val_ratio: float = 0.15, seed: int = 42):
    """
    Split a pool of survey years into TRAIN and VALIDATION, stratified by
    (YEAR, has_MILCO, has_NOMBO) so class & year proportions are preserved.
    """
    rng = np.random.default_rng(seed)
    df_pool = df_pool.copy()
    df_pool["STRATA"] = (
        df_pool["YEAR"] + "_" +
        (df_pool["MILCO"] > 0).astype(int).astype(str) + "_" +
        (df_pool["NOMBO"] > 0).astype(int).astype(str)
    )

    val_indices = []
    for _, group in df_pool.groupby("STRATA"):
        idx = group.index.to_numpy()
        n_val = int(round(len(idx) * val_ratio))
        if n_val == 0 and len(idx) >= 4:
            n_val = 1
        chosen = rng.choice(idx, size=n_val, replace=False)
        val_indices.extend(chosen.tolist())

    df_val = df_pool.loc[df_pool.index.isin(val_indices)].copy()
    df_train = df_pool.loc[~df_pool.index.isin(val_indices)].copy()
    return df_train, df_val


def build_experiment(
    name: str,
    train_years: list,
    test_years: list,
    val_ratio: float = 0.15,
    seed: int = 42
):
    splits_dir = DATASET_DIR / "splits"
    splits_dir.mkdir(parents=True, exist_ok=True)

    df_pool = df_images[df_images["YEAR"].isin(train_years)]
    df_test = df_images[df_images["YEAR"].isin(test_years)]

    df_train, df_val = stratified_year_split(df_pool, val_ratio=val_ratio, seed=seed)

    # Write image path manifest .txt files for Ultralytics YOLO
    train_txt = splits_dir / f"{name}_train.txt"
    val_txt   = splits_dir / f"{name}_val.txt"
    test_txt  = splits_dir / f"{name}_test.txt"

    train_txt.write_text("\n".join(df_train["IMAGE_PATH"].tolist()) + "\n", encoding="utf-8")
    val_txt.write_text("\n".join(df_val["IMAGE_PATH"].tolist()) + "\n", encoding="utf-8")
    test_txt.write_text("\n".join(df_test["IMAGE_PATH"].tolist()) + "\n", encoding="utf-8")

    # Write Ultralytics YAML config
    yaml_path = DATASET_DIR / f"{name}.yaml"
    yaml_content = f"""# {name.upper()} Configuration
path: {DATASET_DIR.resolve()}
train: {train_txt.resolve()}
val: {val_txt.resolve()}
test: {test_txt.resolve()}

nc: 2
names:
  0: MILCO
  1: NOMBO
"""
    yaml_path.write_text(yaml_content, encoding="utf-8")

    # Print verification summary table
    print(f"================================================================================")
    print(f"EXPERIMENT: {name.upper()}")
    print(f"YAML Path : {yaml_path.resolve()}")
    print(f"================================================================================")
    header = f"{'SPLIT':<14}{'SOURCE YEARS':<24}{'IMAGES':<8}{'EMPTY':<8}{'MILCO':<8}{'NOMBO':<8}{'AVG_OBJ':<8}"
    print(header)
    print("-" * len(header))

    for split_label, yrs_str, sub in [
        ("TRAIN (85%)", "+".join(train_years), df_train),
        ("VAL (15%)",   "+".join(train_years), df_val),
        ("TEST (100%)", "+".join(test_years),  df_test),
    ]:
        n_img = len(sub)
        n_emp = int(sub["EMPTY"].sum())
        n_mil = int(sub["MILCO"].sum())
        n_nom = int(sub["NOMBO"].sum())
        avg_o = (n_mil + n_nom) / n_img if n_img > 0 else 0.0
        print(f"{split_label:<14}{yrs_str:<24}{n_img:<8}{n_emp:<8}{n_mil:<8}{n_nom:<8}{avg_o:<8.2f}")
    print("\n")


# 1. Build Primary Experiment (Train/Val: 2010+2015+2017 -> Test: 2018)
build_experiment(
    name="primary_experiment",
    train_years=["2010", "2015", "2017"],
    test_years=["2018"],
    val_ratio=VAL_RATIO,
    seed=RANDOM_SEED
)

# 2. Build Secondary Stress Test (Train/Val: 2010+2015+2017+2018 -> Test: 2021)
build_experiment(
    name="secondary_stress_test",
    train_years=["2010", "2015", "2017", "2018"],
    test_years=["2021"],
    val_ratio=VAL_RATIO,
    seed=RANDOM_SEED
)
```

```python
from pathlib import Path
from PIL import Image
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==============================================================================
# 1. CONFIGURATION & PATHS
# ==============================================================================
DATASET_DIR = Path("./milco_dataset")
PLOTS_DIR = DATASET_DIR / "plots"
PLOTS_DIR.mkdir(parents=True, exist_ok=True)

index_csv = DATASET_DIR / "dataset_image_index.csv"
if not index_csv.exists():
    raise FileNotFoundError(f"Missing {index_csv}. Please run the dataset builder cell first.")

df_images = pd.read_csv(index_csv)
df_images["YEAR"] = df_images["YEAR"].astype(str)

CLASS_MAP = {0: "MILCO", 1: "NOMBO"}

# Standard small-object detection scale thresholds (in pixels^2):
# Tiny: < 16x16 (256 px^2) | Small: 16x16 to 32x32 (1024 px^2) |
# Medium: 32x32 to 96x96 (9216 px^2) | Large: >= 96x96 (9216 px^2)
def categorize_target_size(area_px: float) -> str:
    if area_px < 16 ** 2:
        return "Tiny (<16² px)"
    elif area_px < 32 ** 2:
        return "Small (16²–32² px)"
    elif area_px < 96 ** 2:
        return "Medium (32²–96² px)"
    else:
        return "Large (≥96² px)"

SIZE_ORDER = [
    "Tiny (<16² px)",
    "Small (16²–32² px)",
    "Medium (32²–96² px)",
    "Large (≥96² px)"
]

# ==============================================================================
# 2. EXTRACT ALL BOUNDING BOXES & COMPUTE THE 7 METRICS
# ==============================================================================
bbox_records = []

for _, row in df_images.iterrows():
    img_path = Path(row["IMAGE_PATH"])
    lbl_path = Path(row["LABEL_PATH"])
    year = str(row["YEAR"])

    if not lbl_path.is_file():
        continue

    lines = [
        l.strip() for l in lbl_path.read_text(encoding="utf-8", errors="ignore").splitlines()
        if l.strip()
    ]
    if not lines:
        continue

    # Read native image width and height rapidly from image header
    with Image.open(img_path) as im:
        img_w, img_h = im.size
    img_area = float(img_w * img_h)

    for obj_idx, line in enumerate(lines):
        parts = line.split()
        if len(parts) < 5:
            continue

        cls_id = int(parts[0])
        x_c, y_c, rel_w, rel_h = map(float, parts[1:5])

        # Clamp normalized values safely to (0, 1]
        rel_w = max(1e-6, min(1.0, rel_w))
        rel_h = max(1e-6, min(1.0, rel_h))

        # Calculate the 7 requested metrics:
        bbox_w = rel_w * img_w                     # 1. Bounding-box width (px)
        bbox_h = rel_h * img_h                     # 2. Bounding-box height (px)
        bbox_area = bbox_w * bbox_h                # 3. Bounding-box area (px^2)
        w_ratio = rel_w                            # 4. Width / image width
        h_ratio = rel_h                            # 5. Height / image height
        area_ratio = bbox_area / img_area          # 6. Area / image area
        aspect_ratio = bbox_w / bbox_h             # 7. Aspect ratio (W / H)

        # Also compute equivalent area at standard YOLO imgsz=640x640
        area_at_640 = area_ratio * (640.0 * 640.0)

        bbox_records.append({
            "YEAR": year,
            "IMAGE_NAME": row["IMAGE_NAME"],
            "CLASS_ID": cls_id,
            "CLASS_NAME": CLASS_MAP.get(cls_id, f"Class_{cls_id}"),
            "IMG_WIDTH": img_w,
            "IMG_HEIGHT": img_h,
            "IMG_AREA": img_area,
            "BBOX_WIDTH": bbox_w,
            "BBOX_HEIGHT": bbox_h,
            "BBOX_AREA": bbox_area,
            "WIDTH_RATIO": w_ratio,
            "HEIGHT_RATIO": h_ratio,
            "AREA_RATIO": area_ratio,
            "AREA_PCT": area_ratio * 100.0,
            "ASPECT_RATIO": aspect_ratio,
            "SIZE_CATEGORY": categorize_target_size(bbox_area),
            "SIZE_CATEGORY_640": categorize_target_size(area_at_640)
        })

df_bboxes = pd.DataFrame(bbox_records)
bbox_csv_path = DATASET_DIR / "dataset_bbox_analysis.csv"
df_bboxes.to_csv(bbox_csv_path, index=False)

# ==============================================================================
# 3. PRINT SUMMARY TABLE FOR ALL 7 METRICS
# ==============================================================================
metrics_info = [
    ("BBOX_WIDTH",   "Bounding-box width (px)"),
    ("BBOX_HEIGHT",  "Bounding-box height (px)"),
    ("BBOX_AREA",    "Bounding-box area (px²)"),
    ("WIDTH_RATIO",  "Width / image width"),
    ("HEIGHT_RATIO", "Height / image height"),
    ("AREA_RATIO",   "Area / image area"),
    ("ASPECT_RATIO", "Aspect ratio (W / H)")
]

print(f"=========================================================================================")
print(f"BOUNDING-BOX ANALYSIS ACROSS ALL {len(df_bboxes)} OBJECTS (MILCO: {(df_bboxes['CLASS_NAME']=='MILCO').sum()}, NOMBO: {(df_bboxes['CLASS_NAME']=='NOMBO').sum()})")
print(f"=========================================================================================")
print(f"{'METRIC':<28}{'MEAN':>12}{'STD':>12}{'MEDIAN':>12}{'MIN':>12}{'MAX':>12}")
print("-" * 88)
for col, label in metrics_info:
    s = df_bboxes[col]
    fmt = ".4f" if "RATIO" in col and col != "ASPECT_RATIO" else ".2f"
    print(f"{label:<28}{s.mean():>12{fmt}}{s.std():>12{fmt}}{s.median():>12{fmt}}{s.min():>12{fmt}}{s.max():>12{fmt}}")
print("-" * 88)

# ==============================================================================
# 4. GENERATE THE 4 BOUNDING-BOX PLOTS (2x2 PUBLICATION FIGURE)
# ==============================================================================
plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "axes.labelsize": 11,
    "figure.facecolor": "white"
})

fig, axes = plt.subplots(2, 2, figsize=(16, 12), dpi=150)

# ------------------------------------------------------------------------------
# PLOT 1: Bounding-box size distribution (Width vs. Height + Aspect Ratio)
# ------------------------------------------------------------------------------
ax1 = axes[0, 0]
for cls_name, color, marker in [("MILCO", "#e74c3c", "o"), ("NOMBO", "#f39c12", "^")]:
    sub = df_bboxes[df_bboxes["CLASS_NAME"] == cls_name]
    ax1.scatter(
        sub["BBOX_WIDTH"], sub["BBOX_HEIGHT"],
        label=f"{cls_name} (n={len(sub)})",
        color=color, marker=marker, alpha=0.65, edgecolors="black", linewidth=0.4, s=40
    )

# Plot reference aspect ratio line (W = H) and scale boundaries (16px, 32px, 96px)
max_dim = max(df_bboxes["BBOX_WIDTH"].max(), df_bboxes["BBOX_HEIGHT"].max()) * 1.1
ax1.plot([0, max_dim], [0, max_dim], linestyle="--", color="gray", alpha=0.6, label="Aspect Ratio = 1.0 (Square)")
for thresh, lbl in [(16, "16px (Tiny)"), (32, "32px (Small)"), (96, "96px (Medium)")]:
    if thresh < max_dim:
        ax1.axvline(thresh, color="#34495e", linestyle=":", alpha=0.5)
        ax1.axhline(thresh, color="#34495e", linestyle=":", alpha=0.5)

ax1.set_xlim(0, max_dim)
ax1.set_ylim(0, max_dim)
ax1.set_xlabel("Bounding-Box Width (pixels)")
ax1.set_ylabel("Bounding-Box Height (pixels)")
ax1.set_title("1. Bounding-Box Size Distribution (Width vs. Height)")
ax1.legend(loc="upper left", frameon=True)
ax1.grid(True, linestyle="--", alpha=0.4)

# ------------------------------------------------------------------------------
# PLOT 2: MILCO vs NOMBO Box Sizes (Relative Area % & Aspect Ratio Comparison)
# ------------------------------------------------------------------------------
ax2 = axes[0, 1]
milco_area = df_bboxes[df_bboxes["CLASS_NAME"] == "MILCO"]["AREA_PCT"]
nombo_area = df_bboxes[df_bboxes["CLASS_NAME"] == "NOMBO"]["AREA_PCT"]

bp2 = ax2.boxplot(
    [milco_area, nombo_area],
    labels=["MILCO\n(Mine-Like)", "NOMBO\n(Non-Mine)"],
    patch_artist=True,
    widths=0.42,
    showfliers=False
)
colors_cls = ["#e74c3c", "#f39c12"]
for patch, c in zip(bp2["boxes"], colors_cls):
    patch.set_facecolor(c)
    patch.set_alpha(0.65)
for median in bp2["medians"]:
    median.set(color="black", linewidth=2)

# Overlay jittered individual points
rng = np.random.default_rng(42)
for idx, (series, c) in enumerate([(milco_area, "#c0392b"), (nombo_area, "#d68910")], start=1):
    jitter = rng.normal(loc=idx, scale=0.06, size=len(series))
    ax2.scatter(jitter, series, color=c, alpha=0.45, s=18, edgecolors="none")

ax2.set_ylabel("Relative Box Area (% of Image Area)")
ax2.set_title("2. MILCO vs. NOMBO Box Sizes (Area / Image Area %)")
ax2.grid(axis="y", linestyle="--", alpha=0.5)

# Add summary stats text box inside Plot 2
stats_txt = (
    f"MILCO Median Area: {milco_area.median():.3f}% (Mean: {milco_area.mean():.3f}%)\n"
    f"NOMBO Median Area: {nombo_area.median():.3f}% (Mean: {nombo_area.mean():.3f}%)\n"
    f"MILCO Median AR  : {df_bboxes[df_bboxes['CLASS_NAME']=='MILCO']['ASPECT_RATIO'].median():.2f}\n"
    f"NOMBO Median AR  : {df_bboxes[df_bboxes['CLASS_NAME']=='NOMBO']['ASPECT_RATIO'].median():.2f}"
)
ax2.text(
    0.96, 0.95, stats_txt, transform=ax2.transAxes,
    fontsize=9.5, va="top", ha="right",
    bbox=dict(boxstyle="round,pad=0.5", facecolor="#fef9e7", edgecolor="#7f8c8d", alpha=0.9)
)

# ------------------------------------------------------------------------------
# PLOT 3: Box Sizes by Acquisition Year (Exposing Survey Altitude/Scale Shifts)
# ------------------------------------------------------------------------------
ax3 = axes[1, 0]
years_sorted = sorted(df_bboxes["YEAR"].unique())
year_groups = [df_bboxes[df_bboxes["YEAR"] == y]["AREA_PCT"].values for y in years_sorted]

bp3 = ax3.boxplot(
    year_groups,
    labels=[f"{y}\n(n={len(g)})" for y, g in zip(years_sorted, year_groups)],
    patch_artist=True,
    widths=0.48,
    showfliers=False
)
year_palette = ["#3498db", "#9b59b6", "#1abc9c", "#e67e22", "#34495e"]
for patch, c in zip(bp3["boxes"], year_palette[:len(years_sorted)]):
    patch.set_facecolor(c)
    patch.set_alpha(0.65)
for median in bp3["medians"]:
    median.set(color="black", linewidth=2)

for idx, (vals, c) in enumerate(zip(year_groups, year_palette[:len(years_sorted)]), start=1):
    jitter = rng.normal(loc=idx, scale=0.06, size=len(vals))
    ax3.scatter(jitter, vals, color=c, alpha=0.45, s=16, edgecolors="none")

ax3.set_xlabel("Survey Acquisition Year")
ax3.set_ylabel("Relative Box Area (% of Image Area)")
ax3.set_title("3. Bounding-Box Sizes by Acquisition Year")
ax3.grid(axis="y", linestyle="--", alpha=0.5)

# ------------------------------------------------------------------------------
# PLOT 4: Tiny / Small / Medium / Large Target Distribution
# ------------------------------------------------------------------------------
ax4 = axes[1, 1]
x_pos = np.arange(len(SIZE_ORDER))
width = 0.38

milco_counts = [
    len(df_bboxes[(df_bboxes["CLASS_NAME"] == "MILCO") & (df_bboxes["SIZE_CATEGORY"] == cat)])
    for cat in SIZE_ORDER
]
nombo_counts = [
    len(df_bboxes[(df_bboxes["CLASS_NAME"] == "NOMBO") & (df_bboxes["SIZE_CATEGORY"] == cat)])
    for cat in SIZE_ORDER
]

b_m = ax4.bar(x_pos - width/2, milco_counts, width, label="MILCO", color="#e74c3c", edgecolor="black")
b_n = ax4.bar(x_pos + width/2, nombo_counts, width, label="NOMBO", color="#f39c12", edgecolor="black")

ax4.set_xticks(x_pos)
ax4.set_xticklabels(SIZE_ORDER, rotation=12)
ax4.set_ylabel("Number of Objects")
ax4.set_title("4. Tiny / Small / Medium / Large Target Distribution")
ax4.legend(frameon=True)
ax4.grid(axis="y", linestyle="--", alpha=0.5)
max_bar = max(max(milco_counts, default=1), max(nombo_counts, default=1))
ax4.set_ylim(0, max_bar * 1.22)

total_objs = len(df_bboxes)
for bar in list(b_m) + list(b_n):
    h = int(bar.get_height())
    pct = (h / total_objs) * 100 if total_objs > 0 else 0
    ax4.text(
        bar.get_x() + bar.get_width() / 2, h + (max_bar * 0.02),
        f"{h}\n({pct:.1f}%)",
        ha="center", va="bottom", fontsize=9, fontweight="bold"
    )

plt.tight_layout(pad=2.5)
out_plot_path = PLOTS_DIR / "milco_bbox_analysis_4plots.png"
plt.savefig(out_plot_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"\n[✓] Object-level bounding-box CSV saved to: {bbox_csv_path.resolve()}")
print(f"[✓] 4-panel bounding-box figure saved to : {out_plot_path.resolve()}")
```
